# 📋 Extraction de playlist YouTube pour NotebookLM

### 💡 Pourquoi ce notebook ?
Vous suivez une playlist YouTube de cours, de tutoriels ou de vulgarisation technique, et vous bloquez sur un point précis :
- Impossible de joindre le créateur ou l'enseignant pour lui poser votre question.
- Les IA généralistes (ChatGPT, Claude, etc.), sans contexte, répondent souvent à côté ou proposent des méthodes sans rapport avec la progression pédagogique de la vidéo.

**La solution :** Fournir directement la playlist à **Google NotebookLM** en tant que sources. Grâce à son ancrage strict aux données (*grounding*), NotebookLM devient votre tuteur dédié : il répond avec exactitude **en se basant exclusivement sur les explications et exemples du créateur**.

Ce notebook extrait en quelques secondes la liste propre de toutes les URLs de la playlist, prête à être copiée-collée en un seul clic dans NotebookLM.

```python
#@title 1. Extraction des liens YouTube
#@markdown Entrez l'URL de votre playlist YouTube ci-dessous puis exécutez la cellule.

playlist_url = "https://youtube.com/playlist?list=PL9gnSGHSqcnr_DxHsP7AW9ftq0AtAyYqJ" #@param {type:"string"}

!pip install -q yt-dlp

import yt_dlp

ydl_opts = {
    'extract_flat': True,
    'quiet': True,
}

print("⏳ Récupération des URLs de la playlist...")

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    info = ydl.extract_info(playlist_url, download=False)
    
    urls = []
    if 'entries' in info:
        for item in info['entries']:
            if item:
                video_id = item.get('id')
                urls.append(f"https://www.youtube.com/watch?v={video_id}")

total = len(urls)
print(f"✅ {total} vidéos trouvées !")

# Sauvegarde dans un fichier texte
with open("urls_notebooklm.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(urls))

print("\n" + "="*50)
print("👇 COPIEZ LA LISTE CI-DESSOUS POUR NOTEBOOKLM 👇")
print("="*50 + "\n")
print("\n".join(urls))
print("\n" + "="*50)
print("📁 L'intégralité des URLs a également été enregistrée dans 'urls_notebooklm.txt'.")
```

# 📋 Extraction de playlist YouTube pour NotebookLM

Ce notebook permet d'extraire rapidement la liste des liens de toutes les vidéos d'une **playlist YouTube** afin de les importer en une seule passe comme sources dans **Google NotebookLM**.

> **Rappel :** NotebookLM autorise jusqu'à 50 sources par carnet.

```python
#@title 1. Extraction des liens YouTube
#@markdown Entrez l'URL de votre playlist YouTube et cochez si vous souhaitez limiter l'affichage aux 50 premières sources (limite NotebookLM).

playlist_url = "https://youtube.com/playlist?list=PL9gnSGHSqcnr_DxHsP7AW9ftq0AtAyYqJ" #@param {type:"string"}
limiter_a_50 = True #@param {type:"boolean"}

!pip install -q yt-dlp

import yt_dlp

ydl_opts = {
    'extract_flat': True,
    'quiet': True,
}

print("⏳ Récupération des URLs de la playlist...")

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    info = ydl.extract_info(playlist_url, download=False)
    
    urls = []
    if 'entries' in info:
        for item in info['entries']:
            if item:
                video_id = item.get('id')
                urls.append(f"https://www.youtube.com/watch?v={video_id}")

total = len(urls)
print(f"✅ {total} vidéos trouvées !")

if total > 50 and limiter_a_50:
    print("ℹ️ Limitation active : affichage des 50 premières URLs (limite maximale d'un carnet NotebookLM).")
    urls_a_afficher = urls[:50]
else:
    urls_a_afficher = urls

# Sauvegarde dans un fichier texte
with open("urls_notebooklm.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(urls))

print("\n" + "="*50)
print("👇 COPIEZ LA LISTE CI-DESSOUS POUR NOTEBOOKLM 👇")
print("="*50 + "\n")
print("\n".join(urls_a_afficher))
print("\n" + "="*50)
print("📁 L'intégralité des URLs a également été enregistrée dans 'urls_notebooklm.txt'.")
```

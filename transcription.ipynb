# 🎙️ Transcription de cours avec Whisper (OpenAI)

Ce notebook permet de transcrire automatiquement un enregistrement audio ou vidéo (cours, TD, TP) stocké sur votre **Google Drive** à l'aide du modèle Whisper d'OpenAI.

```python
#@title 1. Installation de Whisper et Connexion à Google Drive
#@markdown Cliquez sur le bouton Exécuter pour préparer l'environnement.

!pip install -q git+https://github.com/openai/whisper.git

from google.colab import drive
drive.mount('/content/drive')

print("✅ Environnement prêt et Google Drive connecté !")
```

```python
#@title 2. Paramètres et Lancement de la Transcription
#@markdown 1. Dans le volet de gauche, cliquez sur l'icône **Dossier** 📁.
#@markdown 2. Naviguez dans `drive` > `MyDrive` vers votre fichier audio/vidéo.
#@markdown 3. Faites un **clic droit sur le fichier** > **Copier le chemin d'accès**.
#@markdown 4. Collez le chemin ci-dessous puis exécutez la cellule.

chemin_fichier = "/content/drive/MyDrive/Enregistrements/MonCours.mp4" #@param {type:"string"}
modele = "medium" #@param ["tiny", "base", "small", "medium", "large"]
langue = "French" #@param ["French", "English", "Auto"]

lang_param = f"--language {langue}" if langue != "Auto" else ""

!whisper "{chemin_fichier}" --model {modele} {lang_param}
```
